```python
data['year_added'] = data['date_added'].dt.year
```

```python
year_added_count = data['year_added'].value_counts().sort_index()

year_added_count
```

```python
fig = px.line(
    x=year_added_count.index,
    y=year_added_count.values,
    markers=True,
    labels={
        'x': 'Year',
        'y': 'Number of Titles Added'
    },
    title='Netflix Titles Added by Year'
)

fig.show()
```

**Inference:** The number of titles added to Netflix increased substantially over time, reaching a peak of **1,999 titles in 2019 (22.69%)**. Additions grew from **73 titles in 2015 to 1,999 in 2019**, an increase of **1,926 titles (2,638%)**. After 2019, additions declined to **1,878 in 2020** and **1,498 in 2021**, representing a **25.06% decrease** from the 2019 peak. Overall, **92.70% of the titles in the dataset were added between 2017 and 2021**.

```python
type_year = pd.crosstab(
    data['year_added'],
    data['type']
)

type_year
```

```python
fig = px.line(
    type_year,
    x=type_year.index,
    y=type_year.columns,
    markers=True,
    title='Movies vs TV Shows Added by Year',
    labels={
        'value': 'Number of Titles',
        'year_added': 'Year',
        'variable': 'Content Type'
    }
)

fig.show()
```

### Inference

* **Movies strongly dominated** yearly additions, especially after 2016.
* Movie additions rose from **253 in 2016 to 1,424 in 2019**, a **463.2% increase**.
* TV Shows also grew from **165 to 575** during the same period, a **248.5% increase**.
* Additions declined after the peak, but Movies still remained **nearly 2×** TV Shows in 2021.

**Overall:** Netflix's content expansion grew rapidly after 2015, with **Movies driving most of the growth**.

```python
movies = data[data['type'] == 'Movie'].copy()
tv_shows = data[data['type'] == 'TV Show'].copy()

movies['duration_min'] = (
    movies['duration']
    .str.extract(r'(\d+)')
    .astype(float)
)

tv_shows['seasons'] = (
    tv_shows['duration']
    .str.extract(r'(\d+)')
    .astype(float)
)
```

```python
fig = px.histogram(
    movies,
    x='duration_min',
    nbins=30,
    title='Distribution of Movie Durations',
    labels={
        'duration_min': 'Duration (Minutes)',
        'count': 'Number of Movies'
    }
)

fig.show()
```

```python
fig = px.histogram(
    tv_shows,
    x='seasons',
    nbins=17,
    title='Distribution of TV Show Seasons',
    labels={
        'seasons': 'Number of Seasons',
        'count': 'Number of TV Shows'
    }
)

fig.show()
```

```python
movie_stats = movies['duration_min'].agg(
    ['count', 'mean', 'median', 'min', 'max']
)

tv_stats = tv_shows['seasons'].agg(
    ['count', 'mean', 'median', 'min', 'max']
)

print("Movie Duration:")
print(movie_stats)

print("\nTV Show Seasons:")
print(tv_stats)
```

### Inference

**Movies have a median duration of 98 minutes**, with an average of **99.56 minutes**, while **TV Shows have a median of 1 season**. The longest movie is **312 minutes**, whereas the longest TV Show has **17 seasons**, showing that most movies are around 1–2 hours while most TV Shows have relatively few seasons.

```python
genre_type = data[['type', 'listed_in']].copy()

genre_type['genre'] = genre_type['listed_in'].str.split(', ')
genre_type = genre_type.explode('genre')

genre_type_count = (
    genre_type
    .groupby(['type', 'genre'])
    .size()
    .reset_index(name='count')
)

fig = px.sunburst(
    genre_type_count,
    path=['type', 'genre'],
    values='count',
    title='Netflix Content Distribution by Type and Genre'
)

fig.update_layout(
    width=1000,
    height=800,
    title_font_size=21
)

fig.update_traces(
    textfont=dict(size=18)
)

fig.show()
```

2. After splitting the multi-genre `listed_in` column, Movies produce **13,190 Movie–Genre associations**, averaging approximately **2.15 genres per movie**.
3. **International Movies (2,752)**, **Dramas (2,427)**, and **Comedies (1,674)** are the most frequent genre categories.
4. International Movies occur **325 more times than Dramas** and **1,078 more times than Comedies**.
5. TV Shows show strong representation in categories such as **International TV Shows, TV Dramas, TV Comedies, and Crime TV Shows**.
6. Overall, the Sunburst highlights the **multi-genre nature of Netflix content**, as individual titles can belong to multiple genres.


## Final Overall Inference

1. **Movies dominate the Netflix catalog**, with **6,131 titles (69.62%)**, while TV Shows account for **2,676 titles (30.38%)**.

2. Netflix experienced a **strong increase in content additions over time**. The number of titles added rose from **73 in 2015 to 1,999 in 2019**, an increase of approximately **2,638%**.

3. After reaching the 2019 peak, content additions declined to **1,878 in 2020** and **1,498 in 2021**, representing a **25.06% decrease from 2019 to 2021**.

4. **TV-MA is the most common rating**, with **3,207 titles (36.41%)**, followed by **TV-14 with 2,160 titles (24.53%)**. Together, these two ratings account for **60.94%** of the entire catalog.

5. The **United States has the highest number of country associations (3,689)**, followed by **India (1,046)** and the **United Kingdom (804)**. The U.S. has approximately **3.53 times** as many country associations as India.

6. **International Movies (2,752)** are the most frequent genre occurrence, followed by **Dramas (2,427)** and **Comedies (1,674)**. International Movies occur **325 more times than Dramas**.

7. The catalog has a strong **multi-genre structure**. The **6,131 movies generate 13,190 Movie–Genre associations**, giving an average of approximately **2.15 genre associations per movie**.

8. Movie duration is relatively concentrated around the typical feature-film length, with a **median of 98 minutes** and an average of approximately **99.56 minutes**. The observed range is **3 to 312 minutes**.

9. TV Shows generally have shorter season counts, with a **median of 1 season**, while the maximum observed value is **17 seasons**.

10. The analysis of **release year** shows that Netflix's catalog contains substantial representation of more recent content, with the strongest concentration of releases occurring during the **2010s and early 2020s**.

11. The comparison of **rating and content type** shows that TV-MA represents a larger share of TV Shows (**42.79%**) than Movies (**33.63%**), indicating a difference in rating composition between the two content types.

12. Overall, the Netflix dataset represents a **movie-heavy and internationally diverse catalog**, with substantial mature-rated content, strong growth in content additions during the late 2010s, and a broad multi-genre structure. The analysis also shows that **Movies and TV Shows differ not only in quantity but also in their genre, rating, duration, and release patterns**.
